```python
import sys
sys.path.append("..")   # geocoding_ml/ -> lets us import from helpers/
import numpy as np
from helpers.pairs import load_pairs

DATASET = "overture_case2s"
train, train_label = load_pairs(DATASET, "train")
val, val_label = load_pairs(DATASET, "val")
test, test_label = load_pairs(DATASET, "test")
train.shape, val.shape, test.shape
```

```python
for df in [train, val, test]:
    df["address_a_tok"] = df["address_a"].apply(lambda x: tok.tokenize(x))
    df["address_b_tok"] = df["address_b"].apply(lambda x: tok.tokenize(x))
    df["pair_length"] = df["address_a_tok"].str.len() + df["address_b_tok"].str.len()
    display(df["pair_length"].describe(percentiles = [0.99]))
```
